```python
curatedObj = eval(str(dbutils.widgets.get('CuratedTables')))
curatedObj
```

```python
newCuratedObj = []

for i in curatedObj:
    newKey = {}
    for x in i:
        newKey[x] = i[x]
    newKey['curatedTable'] = 'candpc.cls_propsecting'
    newCuratedObj.append(newKey)


```

```python
finalCuratedObj = curatedObj + newCuratedObj
finalCuratedObj
```

```python
curatedTables = list(set([i.get('curatedTable') for i in finalCuratedObj if 'curatedTable' in i]))
curatedTables = list(set(curatedTables))
curatedTables
```
