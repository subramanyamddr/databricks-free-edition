```python
data = [
    (101, "Rahul Sharma", "Mumbai", "Maharashtra", "2024-01-01"),
    (101, "Rahul Sharma", "Mumbai", "Maharashtra", "2024-02-10"),
    (102, "Priya Verma", "Delhi", "Delhi", "2024-01-05"),
    (103, "Amit Singh", "Pune", "Maharashtra", "2024-03-01"),
    (103, "Amit Singh", "Pune", "Maharashtra", "2024-02-15")
]

columns = ["customer_id", "name", "city", "state", "update_date"]

df = spark.createDataFrame(data, columns)
display(df)
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window

window_spec = Window.partitionBy('customer_id').orderBy(desc('update_date'))

df_res = df.withColumn('rank_no',rank().over(window_spec)).filter(col('rank_no')==1).drop('rank_no')

display(df_res)
```

another efficient way without rank functions


```python
df_res = df.groupBy("customer_id", "name", "city", "state") \
    .agg(max("update_date").alias("update_date"))

display(df_res)
```

```python
df_res = df.sort(desc("update_date")).dropDuplicates(["customer_id"])

display(df_res)
```
