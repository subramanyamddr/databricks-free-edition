```python
df_usa = df.withColumn('regex_exp', regexp_replace('phone', "[^0-9+]", "")) \
    .withColumn('valid_usa', when(col('regex_exp').rlike(r"^\+1\d{10}$"), col('regex_exp')).otherwise(lit(""))) \
    .select('customer_id', 'name', 'phone', 'valid_usa')

display(df_usa)
```
