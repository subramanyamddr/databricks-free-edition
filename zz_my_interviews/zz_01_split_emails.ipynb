```python
mail_ids = ["alice@example.com", "bob@example.com", "carol@example.com"]
df_mail = spark.createDataFrame([(mail,) for mail in mail_ids], ["mail_id"])
display(df_mail)
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *


df_regex = df_mail.withColumn('split_email',split(col("mail_id"),"@")[1])
df_regex.show()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *


df_regexp_extract = df_mail.withColumn('split_email',regexp_extract(col("mail_id"),"@(.+)",1))
df_regexp_extract.show()
```
