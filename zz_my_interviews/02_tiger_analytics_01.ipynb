```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

df_salary_year = emp_salary_df.withColumn('year', col('month').substr(1, 4))\
                              .groupBy(col('emp_id'),col('year'))\
                              .agg(sum("salary").alias('total_salary'))
                              
df_salary_year.show()
```

```python
emp_df.printSchema()
df_salary_year.printSchema()
```

```python
df_result = emp_df.join(df_salary_year, on="emp_id", how="left_outer")\
                  .filter((col('total_salary')>5000) | col('total_salary').isNull())

df_result.show()
```
