```python
data = [
    ('india','aus','india'),
    ('pak','india','india'),
    ('aus','pak','aus')
]

source_df = spark.createDataFrame(data,['team1','team2','winner'])

display(source_df)
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python


#finding for team1

team1_df = (
    source_df.select(
        col("team1").alias("team"),
        when(col("team1") == col("winner"), 1).otherwise(0).alias("win")
    )
)
team1_df.show()

#finding for team2
team2_df = (
    source_df.select(
        col("team2").alias("team"),
        when(col("team2") == col("winner"), 1).otherwise(0).alias("win")
    )
)
team2_df.show()

#joining team1 and team2
all_teams = team1_df.union(team2_df)
all_teams.show()


# Calculate matches, wins and losses
result = (
    all_teams
    .groupBy("team")
    .agg(
        count("*").alias("matches"),
        sum("win").alias("wins")
    )
    .withColumn("lost", col("matches") - col("wins"))
)

display(result)
```
