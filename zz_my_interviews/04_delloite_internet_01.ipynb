```python
from pyspark.sql import SparkSession

# Create Spark session
spark = SparkSession.builder.appName("CreateDataFrame").getOrCreate()

# Create Spark DataFrame
df = spark.createDataFrame(
    [
        ("John Doe",),
        ("Mary Jane Watson",),
        (None,),
        (" Alice ",),
        ("bob",)
    ],
    ["full_name"]
)

# Show DataFrame
df.show(truncate=False)
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

df_res = df.withColumn('split_col',split(col("full_name")," "))\
           .withColumn('first_name',coalesce(get('split_col',0), lit('N/A')))\
           .withColumn('middle_name',coalesce(get('split_col',1), lit('N/A')))\
            .withColumn('last_name',coalesce(get('split_col',2), lit('N/A')))\
            .withColumn('index_out',coalesce(get('split_col',3), lit('N/A')))
df_res.show()
```
