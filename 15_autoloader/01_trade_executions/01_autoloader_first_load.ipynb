```python
df = spark.readStream\
          .format('cloudFiles')\
          .option('cloudFiles.format','csv')\
          .option('header',True)\
          .option('inferSchema',True)\
          .option('cloudFiles.schemaLocation',"/Volumes/practice/autoloader/autoloader_schema/trade_executions")\
          .load('/Volumes/practice/autoloader/autoloader_input_files/')
```

```python
%sql
create table if not exists practice.autoloader.trade_executions
using delta
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

df.writeStream\
    .format('delta')\
    .option('checkpointLocation','/Volumes/practice/autoloader/autoloader_checkpoint/trade_executions')\
    .option('mergeSchema',True)\
    .trigger(availableNow=True)\
    .table('practice.autoloader.trade_executions')
```
