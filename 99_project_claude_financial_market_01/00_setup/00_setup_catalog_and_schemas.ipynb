```python
%sql
-- =============================================================================
-- 00_setup_catalog_and_schemas.sql
-- Run ONCE per environment by a Unity Catalog admin (cluster-level privilege).
-- Replace <env> with  dev | uat | prod  before executing.
--
-- Run order:
--   1. This file
--   2. 00_setup_volumes.sql
--   3. 00_setup_tables.sql
-- =============================================================================

-- ── 1. Catalog ────────────────────────────────────────────────────────────────
CREATE CATALOG IF NOT EXISTS fin_platform_${env}
  COMMENT 'Financial Platform – ${env} environment';

-- ── 2. Schemas (medallion layers + audit) ────────────────────────────────────
CREATE SCHEMA IF NOT EXISTS fin_platform_${env}.bronze
  COMMENT 'Raw ingest – no transformations, preserves source fidelity';

CREATE SCHEMA IF NOT EXISTS fin_platform_${env}.silver
  COMMENT 'Cleansed, typed and deduplicated data';

CREATE SCHEMA IF NOT EXISTS fin_platform_${env}.gold
  COMMENT 'Business-ready aggregates and feature tables';

CREATE SCHEMA IF NOT EXISTS fin_platform_${env}.audit
  COMMENT 'Pipeline audit, DQ results and lineage';

CREATE SCHEMA IF NOT EXISTS fin_platform_${env}.landing
  COMMENT 'Volume schema for raw file landing zone';

```
