```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

data = [
    (4, "subbu3", "2000")
]

schema = StructType([
    StructField("id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("salary", StringType(), True)
])

df = spark.createDataFrame(data, schema)
display(df)
```

```python
# Dynamic partition overwrite in serverless:
# Instead of spark.conf.set(), use .option("partitionOverwriteMode", "dynamic") on the writer
# This overwrites ONLY the partitions present in the new data, leaving other partitions intact
```

```python
df.write.format('delta').mode('overwrite').option("partitionOverwriteMode", "dynamic").partitionBy('id').saveAsTable('practice.write_mode.emp_table_partition')
```
