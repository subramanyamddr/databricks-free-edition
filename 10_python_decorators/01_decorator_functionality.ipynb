```python
from pyspark.sql.functions import * 
from pyspark.sql.types import * 
from datetime import datetime


def fn_decorator(fn):
    def fn_mainWrapper():
        start_time = datetime.now()
        fn()
        end_time = datetime.now()
        print('start time',start_time)
        print('end time',end_time)
        print('total time',end_time-start_time)
    return fn_mainWrapper

```

```python
@fn_decorator
def fn_readData():
    df = spark.read.format("csv").load('/databricks-datasets/nyctaxi/tripdata/yellow/yellow_tripdata_2009-02.csv.gz')
    return df


display(fn_readData())
```
